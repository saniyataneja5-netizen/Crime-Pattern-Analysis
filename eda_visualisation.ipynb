```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('cleaned_crime_data.csv')
print(df.shape)
df.head()
```

```python
city_counts = df['City'].value_counts().head(15)

plt.figure(figsize=(12, 5))
sns.barplot(x=city_counts.index, y=city_counts.values, palette='Reds_r')
plt.title('Top 15 Cities by Crime Count')
plt.xlabel('City (encoded)')
plt.ylabel('Number of Crimes')
plt.tight_layout()
plt.savefig('plot_city_crime.png')
plt.show()
```

```python
crime_counts = df['Crime Description'].value_counts().head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=crime_counts.values, y=crime_counts.index, palette='Blues_r')
plt.title('Top 10 Crime Types')
plt.xlabel('Count')
plt.tight_layout()
plt.savefig('plot_crime_types.png')
plt.show()
```

```python
plt.figure(figsize=(12, 4))
sns.histplot(df['Time of Occurrence'], bins=24, color='steelblue')
plt.title('Crime Distribution by Time of Day')
plt.xlabel('Time (encoded)')
plt.ylabel('Count')
plt.savefig('plot_time_distribution.png')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.savefig('plot_heatmap.png')
plt.show()
```

```python
closed = df['Case Closed'].value_counts()

plt.figure(figsize=(5, 5))
plt.pie(closed.values, labels=['Open', 'Closed'], autopct='%1.1f%%', colors=['#E24B4A','#1D9E75'])
plt.title('Case Closed vs Open')
plt.savefig('plot_case_status.png')
plt.show()
```

```python
plt.figure(figsize=(10, 4))
sns.histplot(df['Victim Age'], bins=30, kde=True, color='purple')
plt.title('Victim Age Distribution')
plt.xlabel('Age')
plt.savefig('plot_victim_age.png')
plt.show()
```

```python
domain_counts = df['Crime Domain'].value_counts()

plt.figure(figsize=(8, 5))
sns.barplot(x=domain_counts.index, y=domain_counts.values, palette='magma')
plt.title('Crime by Domain/Category')
plt.xlabel('Domain (encoded)')
plt.ylabel('Count')
plt.savefig('plot_crime_domain.png')
plt.show()
```
